```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
sns.boxplot(x=df["fare_amount"])
plt.title("Boxplot of Fare Amount")
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.boxplot(x=df["trip_distance"])
plt.title("Boxplot of Trip Distance")   
plt.show()
```

```python
Q1 = df["fare_amount"].quantile(0.25)
Q3 = df["fare_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
```

```python
fare_outliers = df[
    (df["fare_amount"] < lower) |
    (df["fare_amount"] > upper)
]

print(len(fare_outliers))
```

```python
Q1 = df["trip_distance"].quantile(0.25)
Q3 = df["trip_distance"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
```

```python
trip_distance_outliers = df[
    (df["trip_distance"] < lower) |
    (df["trip_distance"] > upper)
]

print(len(trip_distance_outliers))
```

```python
df[df["fare_amount"] < 100]["fare_amount"].hist(
    bins=50,
    figsize=(10,5)
)
plt.title("Histogram of Fare Amount (Filtered)")
plt.show()
```

```python
df[df["trip_distance"] < 50]["trip_distance"].hist(
    bins=50,
    figsize=(10,5)
)
plt.title("Histogram of Trip Distance (Filtered)")
plt.show()
```
