## Filter African Countries

The World Bank dataset contains all countries and regions.

For this project, only African countries will be retained.

```python
africa = [
    "Kenya","Uganda","Tanzania","Rwanda","Burundi",
    "South Sudan","Ethiopia","Somalia","Djibouti","Eritrea",
    "Nigeria","Ghana","South Africa","Egypt","Morocco",
    "Algeria","Tunisia","Libya","Sudan","Cameroon",
    "Senegal","Zambia","Zimbabwe","Botswana","Namibia",
    "Mozambique","Malawi","Madagascar","Mauritius","Seychelles",
    "Angola","Benin","Burkina Faso","Cape Verde",
    "Central African Republic","Chad","Comoros",
    "Republic of the Congo","DR Congo","Equatorial Guinea",
    "Gabon","Gambia","Guinea","Guinea-Bissau",
    "Ivory Coast","Lesotho","Liberia","Mali",
    "Mauritania","Niger","Sao Tome and Principe",
    "Sierra Leone","Eswatini","Togo"
]
```

```python
df_africa = df[df["Short Name"].isin(africa)]

print(df_africa.shape)

df_africa.head()
```

```python
#Rename column "Indicator Namex" to "Indicator Name" to avoid confusion with the other "Indicator Name" column in the merged dataframe
df_africa = df_africa.rename(columns={"Indicator Name_x": "Indicator Name"})
```

```python
# Checking if the selected indicators are present in the dataset

selected_indicators = [
    "NY.GDP.MKTP.KD.ZG",  # GDP growth (annual %)
    "IT.NET.USER.ZS",     # Internet users (per 100 people)
    "SE.SEC.ENRR"         # Secondary school enrollment (% gross)
]
df_africa[df_africa["Indicator Code"]
          .isin(selected_indicators)][["Indicator Code", "Indicator Name"]].drop_duplicates()
```

```python
print(df_africa.columns.tolist())
```

```python
df_africa[df_africa["Indicator Code"]
          .isin(selected_indicators)].groupby(["Indicator Name"]).size().reset_index(name='count')
```

```python
years = [str(year) for year in range(2000, 2025)]
indicator_data = df_africa[df_africa["Indicator Code"].isin(selected_indicators)]
indicator_data[years].isnull().mean() * 100  # Calculate percentage of missing values for each year
```
